```python
def func(x):
    return (x+3)**2
```

```python
def derivative(x):
    return 2*(x+3)
```

```python
gd=[]
def grad_descent(start, itr, learning_rate):
    x=start
    gd.append(x)
    for i in range(itr):
        x= x-(learning_rate*derivative(x))
        y=func(x)
        print(f"Iteration : {i} with x: {x} and y:{y}")
        gd.append(x)
    return x
```

```python
minval= grad_descent(2, 200, 0.01)
print(f"local min: {minval}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.lineplot(gd)
```
