```python
def func(x):
    return (x+3)**2
```

```python
def derivative(x):
    return 2*(x+3)
```

```python
gd=[]
def gradient_descent(start, itr, learning_rate):
    x= start
    gd.append(x)
    for i in range(itr):
        deri= derivative(x)
        x= x-(learning_rate*deri)
        gd.append(x)
        y= func(x)
        print(f"At iteration {i} the value of x is {x} and value of y is {y}")
    return x
```

```python
start=2
itr=100
learning_rate=0.05
minpoint = gradient_descent(start, itr, learning_rate)
```

```python
print(f"min value : {minpoint}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.lineplot(gd)
plt.title("Gradient descent")
```
